##명륜1동 0세~100세 연령대별 그래프

```python
f=open('/content/wonju_total.csv', encoding ='cp949')
data =csv.reader(f)
##우리동네 리스트
myeong1 = []
for row in data:
    if '강원특별자치도 원주시 명륜1동(5113054100)'  in row[0]:
        for i in row[3:]:
            myeong1.append(int(i.replace(',','')))
```

```python
myeong1
```

```python
plt. plot(myeong1, c = 'r', label='MyeongRyun1')
plt.xlabel("Age", fontsize = 19)
plt.ylabel("Numbers", fontsize = 19)
plt.legend()
plt.show()
```

###원주시 명륜1동의 연령대별 성별 비율 산점도 그리기
1. 명륜1동 성별로 각각 인구데이터 저장
2. 산점도 그리기

```python
f2=open('/content/wonju_gender.csv', encoding ='cp949')
data2 =csv.reader(f2)
male = []
female = []

for row in data2:
    if '명륜1동'  in row[0]:
        for i in row[3:104]:
            male.append(int(i.replace(',', '')))
        for i in row[106:207]:
            female.append(int(i.replace(',', '')))
```

```python
##남녀인구 파이차트
f_w = open('/content/wonju_gender.csv', encoding = 'cp949')
data_w = csv.reader(f_w)

male_w = [] ##원주 남자
female_w = [] ## 원주 여자
#name= input('동네 이름 입력')
for row in data_w:
    if "강원특별자치도 원주시 (5113000000)" in row[0]:
        for i in range(3,104): # Changed range to start from 3 and go up to 103 for 101 age groups
            male_w.append(int(row[i].replace(',', '')))
            female_w.append(int(row[i+103].replace(',', '')))
        break # Added break to stop after finding the correct row
print('원주 : 연령별 남자 =', male_w)
print('원주 : 연령별 여자 = ', female_w)
```
