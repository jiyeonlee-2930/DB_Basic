```python
##서울의 최고 기온 리스트 만들기
f=open('/content/seoul_temp.csv', encoding='cp949')
data = csv.reader(f)
next(data)
##최고기온 리스트 만들기
max_temp=[]
for row in data:
    if row[4]!='':
        max_temp.append(float(row[4]))
print(max_temp)
```

```python
##최고기온 꺽은선 그래프 그리기
plt.figure(figsize = (20,4))
plt.plot(max_temp,color = 'hotpink' )
plt.show()
```

```python
##최고기온 막대 그래프 그리기
plt.figure(figsize = (10,4))
plt.hist(max_temp, bins = 100)
plt.show()
```

```python
plt.boxplot(max_temp)
plt.show()
```

```python
##데이터 월별로 비교
f=open('/content/seoul_temp.csv', encoding='cp949')
data = csv.reader(f)
next(data)
##월별 데이터 리스트
month =[[],[],[],[],[],[],[],[],[],[],[],[]]

for row in data:
    if row[4]!='':
        month[int(row[2].split('-')[1])-1].append(float(row[4]))

plt.boxplot(month)
plt.show()
```

```python
## 원주의 10월중 기온이 가장 높았던 날과 기온 추출
import csv
f=open('/content/wonju_202310.csv', encoding='cp949')
data = csv.reader(f)
next(data)
max_temp=-999
max_date=''
for row in data:
    if row[-1]=='':
        row[-1]= -999
    row[-1] = float(row[-1])
    if max_temp < row[-1]:
        max_date = row[0]
        max_temp = row[-1]
f.close()
print('원주의 10월중 기온이 가장 높았던 날은', max_date,'날', max_temp, '도 입니다.')
```

```python
##2020년 이후 원주의 1월과 2월의 최고기온 추출
import csv
f = open('/content/wonju_temp(10).csv', encoding = 'cp949')
data = csv.reader(f)
next(data)
Jau = []
Feu = []
Mar = []
for row in data:
    if row[-1] != '':
        date = row[0].split('-')
        if 2014 <= int(date[0]) <= 2024:
            if date[1] == '01':
                Jau.append(float(row[-1]))
            elif date[1] == '02':
                Feu.append(float(row[-1]))
            elif date[1] == '03':
                Mar.append(float(row[-1]))
```

```python
import matplotlib.pyplot as plt

plt.plot(Jau, color= 'red')
plt.plot(Feu, color= 'green')
plt.plot(Mar, color= 'pink')
plt.show()
```
